```python
import scanpy as sc
import pandas as pd
import numpy as np

# -----------------------------
# Find Gpr88 column
# -----------------------------
gpr88_idx = np.where(adata.var["gene_symbol"] == "Gpr88")[0][0]

obs_index = pd.Index(adata.obs.index)

results = []

# Search populations related to VMH or CeA
keywords = ["VMH", "CEA"]

for keyword in keywords:

    subclasses = sorted(
        hy_meta.loc[
            hy_meta["subclass"].str.contains(keyword, case=False, na=False),
            "subclass"
        ].unique()
    )

    for subclass in subclasses:

        labels = hy_meta.loc[
            hy_meta["subclass"] == subclass,
            "cell_label"
        ]

        idx = obs_index.get_indexer(labels)
        idx = idx[idx >= 0]

        if len(idx) == 0:
            continue

        values = []

        # Read in chunks (avoids MemoryError)
        for i in range(0, len(idx), 500):

            chunk = idx[i:i+500]

            x = adata[chunk, gpr88_idx].X

            if hasattr(x, "toarray"):
                x = x.toarray().ravel()
            else:
                x = np.asarray(x).ravel()

            values.extend(x)

        values = np.asarray(values)

        results.append({
            "Brain Region": keyword,
            "Neuron Population": subclass,
            "Number of Cells": len(values),
            "Mean GPR88": round(values.mean(),4),
            "Median GPR88": round(np.median(values),4),
            "% Positive Cells": round((values>0).mean()*100,2)
        })

# -----------------------------
# Create dataframe
# -----------------------------
results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    ["Brain Region","Mean GPR88"],
    ascending=[True,False]
)

print(results_df)

# -----------------------------
# Save to Excel
# -----------------------------
results_df.to_csv(
    r"C:\Users\ARCH\Downloads\GPR88_VMH_CeA_Expression.csv",
    index=False
)
```

```python
import os
import numpy as np
import pandas as pd

# =====================================================
# CONFIGURATION
# =====================================================

out_dir = r"D:\Mouse\Results\PAL_Subregions"
os.makedirs(out_dir, exist_ok=True)

# -----------------------------------------------------
# Set to True once you know the correct anatomical column
# -----------------------------------------------------
USE_ANATOMICAL_COLUMN = False

# Example:
# "region_of_interest_acronym"
# "parcellation_structure"
# "parcellation_division"

ANATOMICAL_COL = "parcellation_structure"

# -----------------------------------------------------
# Regions
# -----------------------------------------------------

regions = {
    "CEA": r"\bCEA\b",
    "BST": r"\bBST\b",
    "AAA": r"\bAAA\b",
    "ACB": r"\bACB\b"
}

# =====================================================
# PREPARE
# =====================================================

obs_index = adata.obs.index

matches = np.where(
    adata.var["gene_symbol"] == "Gpr88"
)[0]

if len(matches) == 0:
    raise ValueError("Gpr88 not found in adata.var['gene_symbol'].")

gpr88_idx = matches[0]

# =====================================================
# MAIN LOOP
# =====================================================

for region_name, pattern in regions.items():

    print("\n" + "="*60)
    print(f"Processing {region_name}")
    print("="*60)

    # -------------------------------------------------
    # Region filtering
    # -------------------------------------------------

    if USE_ANATOMICAL_COLUMN:

        region_meta = meta[
            meta[ANATOMICAL_COL] == region_name
        ].copy()

    else:

        region_meta = meta[
            meta["cluster"].str.contains(
                pattern,
                case=False,
                regex=True,
                na=False
            )
        ].copy()

    print("Metadata cells:", len(region_meta))

    if region_meta.empty:
        print("No metadata found.")
        continue

    # -------------------------------------------------
    # Cluster analysis
    # -------------------------------------------------

    results = []

    matched_total = 0

    clusters = sorted(
        region_meta["cluster"].dropna().unique()
    )

    for cluster in clusters:

        subset = region_meta[
            region_meta["cluster"] == cluster
        ]

        labels = subset["cell_label"]

        idx = obs_index.get_indexer(labels)
        idx = idx[idx >= 0]

        matched_total += len(idx)

        if len(idx) < 5:
            continue

        values = []

        for start in range(0, len(idx), 500):

            chunk = idx[start:start+500]

            expr = adata[
                chunk,
                gpr88_idx
            ].X

            if hasattr(expr, "toarray"):
                expr = expr.toarray().ravel()
            else:
                expr = np.asarray(expr).ravel()

            values.extend(expr)

        values = np.asarray(values)

        results.append({

            "Cluster": cluster,

            "Subclass": subset["subclass"].iloc[0],

            "Supertype": subset["supertype"].iloc[0],

            "Cells": len(values),

            "Mean_GPR88": values.mean(),

            "Median_GPR88": np.median(values),

            "Positive_Count": (values > 0).sum(),

            "Positive_%": (values > 0).mean() * 100

        })

    cluster_df = pd.DataFrame(results)

    if cluster_df.empty:
        print("No clusters with >=5 matched cells.")
        continue

    cluster_df = cluster_df.sort_values(
        "Mean_GPR88",
        ascending=False
    ).reset_index(drop=True)

    # -------------------------------------------------
    # Supertype summary
    # -------------------------------------------------

    super_df = (

        cluster_df

        .groupby(
            ["Subclass", "Supertype"],
            as_index=False
        )

        .apply(

            lambda g: pd.Series({

                "Cells":
                    g["Cells"].sum(),

                "Mean_GPR88":
                    (
                        g["Mean_GPR88"] *
                        g["Cells"]
                    ).sum() /
                    g["Cells"].sum(),

                "Representative_Median_GPR88":
                    g["Median_GPR88"].median(),

                "Positive_%":
                    (
                        g["Positive_Count"].sum()
                        /
                        g["Cells"].sum()
                    ) * 100

            })

        )

        .reset_index(drop=True)

        .sort_values(
            "Mean_GPR88",
            ascending=False
        )

    )

    export_cluster = cluster_df.drop(
        columns=["Positive_Count"]
    )

    # -------------------------------------------------
    # Save
    # -------------------------------------------------

    cluster_file = os.path.join(
        out_dir,
        f"{region_name}_Cluster_GPR88.csv"
    )

    super_file = os.path.join(
        out_dir,
        f"{region_name}_Supertype_GPR88.csv"
    )

    export_cluster.to_csv(
        cluster_file,
        index=False
    )

    super_df.to_csv(
        super_file,
        index=False
    )

    print(f"Matched cells: {matched_total}")
    print(f"Clusters saved : {len(export_cluster)}")
    print(f"Supertypes saved : {len(super_df)}")

print("\n" + "="*60)
print("Analysis completed successfully.")
print("Results saved to:")
print(out_dir)
print("="*60)
```

```python
import os
import numpy as np
import pandas as pd

# =====================================================
# CONFIGURATION
# =====================================================

out_dir = r"D:\Mouse\Results\PAL_Subregions"
os.makedirs(out_dir, exist_ok=True)

# -----------------------------------------------------
# Switch to True once anatomical metadata is confirmed
# -----------------------------------------------------
USE_ANATOMICAL_COLUMN = False

# Examples:
# region_of_interest_acronym
# parcellation_structure
# parcellation_division
ANATOMICAL_COL = "parcellation_structure"

# -----------------------------------------------------
# Target regions
# -----------------------------------------------------
regions = {
    "CEA": r"\bCEA\b",
    "BST": r"\bBST\b",
    "AAA": r"\bAAA\b",
    "ACB": r"\bACB\b"
}

# =====================================================
# PREPARE
# =====================================================

obs_index = adata.obs.index

matches = np.where(adata.var["gene_symbol"] == "Gpr88")[0]

if len(matches) == 0:
    raise ValueError("Gpr88 not found in adata.var['gene_symbol'].")

gpr88_idx = matches[0]

# =====================================================
# MAIN LOOP
# =====================================================

for region_name, pattern in regions.items():

    print("\n" + "=" * 70)
    print(f"Processing region: {region_name}")
    print("=" * 70)

    # -------------------------------------------------
    # Region filtering
    # -------------------------------------------------
    if USE_ANATOMICAL_COLUMN:
        region_meta = meta[meta[ANATOMICAL_COL] == region_name].copy()
    else:
        region_meta = meta[
            meta["cluster"].str.contains(
                pattern, case=False, regex=True, na=False
            )
        ].copy()

    print(f"Metadata cells found: {len(region_meta)}")

    if region_meta.empty:
        print("No metadata found for this region. Skipping...")
        continue

    # -------------------------------------------------
    # Cluster analysis
    # -------------------------------------------------
    results = []
    matched_total = 0
    clusters = sorted(region_meta["cluster"].dropna().unique())

    for cluster in clusters:
        subset = region_meta[region_meta["cluster"] == cluster]
        labels = subset["cell_label"]

        # Match metadata labels to h5ad indices
        idx = obs_index.get_indexer(labels)
        idx = idx[idx >= 0]

        matched_total += len(idx)

        # Ignore tiny clusters
        if len(idx) < 5:
            continue

        values = []

        # Read GPR88 expression in chunks
        for start in range(0, len(idx), 500):
            chunk = idx[start : start + 500]
            expr = adata[chunk, gpr88_idx].X

            if hasattr(expr, "toarray"):
                expr = expr.toarray().ravel()
            else:
                expr = np.asarray(expr).ravel()

            values.extend(expr)

        values = np.asarray(values)

        results.append({
            "Cluster": cluster,
            "Subclass": subset["subclass"].iloc[0],
            "Supertype": subset["supertype"].iloc[0],
            "Cells": len(values),
            "Mean_GPR88": values.mean(),
            "Median_GPR88": np.median(values),
            "Positive_Count": (values > 0).sum(),
            "Positive_%": (values > 0).mean() * 100
        })

    cluster_df = pd.DataFrame(results)

    if cluster_df.empty:
        print("No clusters with ≥5 matched cells. Skipping...")
        continue

    cluster_df = cluster_df.sort_values(
        "Mean_GPR88", ascending=False
    ).reset_index(drop=True)

    # -------------------------------------------------
    # Supertype summary (Native Vectorized .agg)
    # -------------------------------------------------
    # Pre-compute expression mass for exact math across all Pandas versions
    cluster_df["Total_GPR88_Mass"] = cluster_df["Mean_GPR88"] * cluster_df["Cells"]

    super_df = (
        cluster_df
        .groupby(["Subclass", "Supertype"], as_index=False)
        .agg(
            Cells=("Cells", "sum"),
            Total_Mass=("Total_GPR88_Mass", "sum"),
            Representative_Median_GPR88=("Median_GPR88", "median"),
            Total_Positive=("Positive_Count", "sum")
        )
    )

    # Mathematical weighted averages
    super_df["Mean_GPR88"] = super_df["Total_Mass"] / super_df["Cells"]
    super_df["Positive_%"] = (super_df["Total_Positive"] / super_df["Cells"]) * 100

    # Clean intermediate calculation columns & sort
    super_df = (
        super_df
        .drop(columns=["Total_Mass", "Total_Positive"])
        .sort_values("Mean_GPR88", ascending=False)
        .reset_index(drop=True)
    )

    # -------------------------------------------------
    # Export cluster table
    # -------------------------------------------------
    export_cluster = cluster_df.drop(
        columns=["Positive_Count", "Total_GPR88_Mass"]
    )

    # Output filenames
    cluster_file = os.path.join(out_dir, f"{region_name}_Cluster_GPR88.csv")
    super_file = os.path.join(out_dir, f"{region_name}_Supertype_GPR88.csv")

    # Save CSVs
    export_cluster.to_csv(cluster_file, index=False)
    super_df.to_csv(super_file, index=False)

    # Logging
    print(f"Matched cells        : {matched_total}")
    print(f"Clusters exported    : {len(export_cluster)}")
    print(f"Supertypes exported  : {len(super_df)}")
    print(f"Saved cluster file   : {cluster_file}")
    print(f"Saved supertype file : {super_file}")

# =====================================================
# END OF SCRIPT
# =====================================================

print("\n" + "=" * 70)
print("✓ PAL GPR88 analysis completed successfully.")
print(f"Results saved in:\n{out_dir}")
print("=" * 70)
```

```python
datasets = {
    "Iso1": iso1,
    "CTXsp": ctx,
    "OLF": olf,
    "PAL": pal
}
import numpy as np
import pandas as pd

dp_types = [
    "0046 IT AON-TT-DP Glut_1",
    "0047 IT AON-TT-DP Glut_2",
    "0048 IT AON-TT-DP Glut_3",
    "0049 IT AON-TT-DP Glut_4",
    "0050 IT AON-TT-DP Glut_5",
    "0402 MPO-ADP Lhx8 Gaba_1",
    "0403 MPO-ADP Lhx8 Gaba_2",
    "0404 MPO-ADP Lhx8 Gaba_3",
    "0405 MPO-ADP Lhx8 Gaba_4",
    "0406 MPO-ADP Lhx8 Gaba_5",
    "0527 ADP-MPO Trp73 Glut_1",
    "0528 ADP-MPO Trp73 Glut_2",
    "0529 ADP-MPO Trp73 Glut_3"
]

gene_symbols = iso1.var["gene_symbol"].values

expr_matrix = pd.DataFrame(index=gene_symbols)

for st in dp_types:

    print("\nProcessing", st)

    X_all = []

    cells = meta.loc[
        meta["supertype"] == st,
        "cell_label"
    ]

    for name, ad in datasets.items():

        common = ad.obs.index.intersection(cells)

        if len(common) == 0:
            continue

        X = ad[common.tolist(), :].X

        if hasattr(X, "toarray"):
            X = X.toarray()

        X_all.append(X)

    if len(X_all) == 0:
        continue

    X_all = np.vstack(X_all)

    expr_matrix[st] = X_all.mean(axis=0)

expr_matrix.index.name = "Gene"

expr_matrix.to_csv("DP_AllGenes_Merged.csv")
```

```python
import pandas as pd
import numpy as np

WORKDIR = r"D:\Mouse\DP_Receptor_Specificity"

###############################################################
# LOAD DATA
###############################################################

df = pd.read_csv(
    fr"{WORKDIR}\Specific_DP_ReceptorSpecificity.csv"
)

if "Unnamed: 0" in df.columns:
    df = df.rename(columns={"Unnamed: 0":"Gene"})

shared = pd.read_csv(
    fr"{WORKDIR}\Specific_DP_CEA_SharedReceptors.csv"
)

if "Unnamed: 0" in shared.columns:
    shared = shared.rename(columns={"Unnamed: 0":"Gene"})

###############################################################
# SHARED DP-CEA
###############################################################

df["Shared_DP_CEA"] = df["Gene"].isin(shared["Gene"])

###############################################################
# DP SPECIFICITY
###############################################################

df["DP_Specificity"] = (
    df["DP_vs_Neighbour_log2FC"]
    - (
        df["DP_vs_CEA_log2FC"] +
        df["DP_vs_BLA_log2FC"]
    )/2
)

###############################################################
# EXPRESSION SCORE
###############################################################

df["ExpressionScore"] = (
    df["DP_Mean"]/
    df["DP_Mean"].max()
)*10

###############################################################
# DRUGGABILITY
###############################################################

druggable = {

"Grm1":"High",
"Grm3":"High",
"Grm5":"High",

"Chrm1":"High",
"Chrm3":"High",

"Drd1":"High",
"Drd2":"High",
"Drd5":"Moderate",

"Htr2a":"High",

"Adra1a":"High",
"Adra1d":"Moderate",
"Adrb2":"High",

"Gpr4":"Moderate",
"Gpr88":"Moderate",

"Cnr1":"High",

"Gabbr1":"High",
"Gabbr2":"High",

"Ntrk2":"Moderate",

"Grin1":"Moderate",
"Grin2a":"Moderate",
"Grin2c":"Moderate",

"Gria1":"Moderate",
"Gria4":"Moderate"

}

df["Druggability"] = (
    df["Gene"]
    .map(druggable)
    .fillna("Unknown")
)

###############################################################
# PRIORITY
###############################################################

def priority(row):

    if (
        row["Shared_DP_CEA"]
        and
        row["DP_Specificity"] > 0
        and
        row["Druggability"]=="High"
    ):
        return "Tier 1"

    elif (
        row["Shared_DP_CEA"]
        and
        row["DP_Specificity"] > 0
    ):
        return "Tier 2"

    elif row["DP_Specificity"] > 0:
        return "Tier 3"

    else:
        return "Low"

df["Priority"] = df.apply(priority,axis=1)

###############################################################
# SORT
###############################################################

df = df.sort_values(
    by=[
        "Priority",
        "DP_Specificity",
        "ExpressionScore"
    ],
    ascending=[True,False,False]
)

###############################################################
# SAVE
###############################################################

outfile = fr"{WORKDIR}\Final_Receptor_Prioritization.csv"

df.to_csv(
    outfile,
    index=False
)

###############################################################
# DISPLAY
###############################################################

print(df[[
    "Gene",
    "Priority",
    "Shared_DP_CEA",
    "DP_Mean",
    "CEA_Mean",
    "DP_Specificity",
    "ExpressionScore",
    "Druggability"
]].head(50))

print("\nSaved to:")
print(outfile)
```
